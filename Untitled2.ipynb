```python
import matplotlib.pyplot as plt

# FIND-S Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

hypothesis = ['0'] * 6

for row in data:
    if row[-1] == 'Yes':
        for i in range(6):
            if hypothesis[i] == '0':
                hypothesis[i] = row[i]
            elif hypothesis[i] != row[i]:
                hypothesis[i] = '?'

print("FIND-S Algorithm")
print("Final Hypothesis:", hypothesis)

# Display graphical output
features = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast']

plt.figure(figsize=(9, 4))
plt.axis('off')

table = plt.table(
    cellText=[hypothesis],
    colLabels=features,
    loc='center',
    cellLoc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1, 2)

plt.title("FIND-S Final Hypothesis")
plt.savefig("Exp-1-Output.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

S = ['0'] * 6
G = [['?'] * 6]

for row in data:
    x = row[:-1]
    label = row[-1]
    if label == 'Yes':
        # Initialize S with first positive example if it's still ['0']*6
        if S == ['0'] * 6:
            S = list(x)
        else:
            for i in range(6):
                if S[i] != x[i]:
                    S[i] = '?'
        # Filter general hypotheses
        G = [g for g in G if all(g[i] == '?' or g[i] == S[i] for i in range(6))]
    else:
        new_G = []
        for g in G:
            for i in range(6):
                if g[i] == '?':
                    if S[i] != '?' and S[i] != x[i]:
                        g_new = list(g)
                        g_new[i] = S[i]
                        if g_new not in new_G:
                            new_G.append(g_new)
        G = [g for g in new_G if any(g[i] != '?' and g[i] == S[i] for i in range(6))]

if not G:
    G = [['?'] * 6]

print("Specific Boundary:", S)
print("General Boundary:", G)

# Prepare visualization
plt.figure(figsize=(10, 4))
plt.axis('off')

cell_text = [S]
row_labels = ['S Boundary']
for idx, g in enumerate(G):
    cell_text.append(g)
    row_labels.append(f'G Boundary {idx+1}' if len(G) > 1 else 'G Boundary')

table = plt.table(
    cellText=cell_text,
    rowLabels=row_labels,
    colLabels=['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast'],
    loc='center',
    cellLoc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 2)

plt.title("Candidate Elimination Boundaries")
plt.savefig("Exp-2-Output.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
import math
import matplotlib.pyplot as plt

data = [
    ['Sunny', 'Hot', 'No'],
    ['Sunny', 'Cool', 'Yes'],
    ['Rainy', 'Hot', 'Yes'],
    ['Rainy', 'Cool', 'Yes'],
    ['Overcast', 'Hot', 'Yes'],
    ['Overcast', 'Cool', 'Yes']
]

def entropy(rows):
    total = len(rows)
    counts = [sum(1 for r in rows if r[-1] == c) for c in ['Yes', 'No']]
    result = 0
    for count in counts:
        if count:
            p = count / total
            result -= p * math.log2(p)
    return result

def id3(rows, features):
    if all(r[-1] == rows[0][-1] for r in rows):
        return rows[0][-1]

    if not features:
        return max(['Yes', 'No'], key=lambda c: sum(r[-1] == c for r in rows))

    gains = []
    for feature in features:
        index = feature
        values = set(r[index] for r in rows)
        remainder = sum(
            (len([r for r in rows if r[index] == v]) / len(rows)) * entropy([r for r in rows if r[index] == v])
            for v in values
        )
        gains.append((entropy(rows) - remainder, index))

    best = max(gains)[1]
    tree = {}

    for value in set(r[best] for r in rows):
        subset = [r for r in rows if r[best] == value]
        tree[value] = id3(subset, [f for f in features if f != best])

    return {best: tree}

tree = id3(data, [0, 1])
print("Decision Tree:", tree)

fig, ax = plt.subplots(figsize=(8, 5))
ax.axis('off')

ax.text(0.5, 0.9, "Outlook", ha='center', bbox=dict(boxstyle='round', fc='lightblue'))

positions = [(0.2, 0.55), (0.5, 0.55), (0.8, 0.55)]
labels = ['Sunny', 'Rainy', 'Overcast']

for pos, label in zip(positions, labels):
    ax.text(*pos, label, ha='center', bbox=dict(boxstyle='round', fc='lightgreen'))
    ax.plot([0.5, pos[0]], [0.87, pos[1] + 0.05], 'k-')

ax.text(0.2, 0.25, 'Decision: Yes / No', ha='center')
ax.text(0.5, 0.25, 'Decision: Yes', ha='center')
ax.text(0.8, 0.25, 'Decision: Yes', ha='center')

plt.title("ID3 Decision Tree")
plt.savefig("Exp-3-Output.png")
plt.show()
```

```python
import random
import matplotlib.pyplot as plt

random.seed(1)

X = [[0, 0], [0, 1], [1, 0], [1, 1]]
Y = [0, 1, 1, 0]

def sigmoid(x):
    return 1 / (1 + 2.71828 ** (-x))

w1 = [[random.uniform(-1, 1) for _ in range(2)] for _ in range(2)]
w2 = [random.uniform(-1, 1) for _ in range(2)]
b1 = [0.0, 0.0]
b2 = 0.0
lr = 0.5
losses = []

for epoch in range(5000):
    total_loss = 0
    for x, target in zip(X, Y):
        h = [sigmoid(sum(x[i] * w1[j][i] for i in range(2)) + b1[j]) for j in range(2)]
        output = sigmoid(sum(h[j] * w2[j] for j in range(2)) + b2)
        error = output - target
        total_loss += error ** 2

        delta2 = error * output * (1 - output)
        delta1 = [delta2 * w2[j] * h[j] * (1 - h[j]) for j in range(2)]

        for j in range(2):
            w2[j] -= lr * delta2 * h[j]
            b1[j] -= lr * delta1[j]
            for i in range(2):
                w1[j][i] -= lr * delta1[j] * x[i]
        b2 -= lr * delta2

    losses.append(total_loss / 4)

print("XOR Predictions:")
for x in X:
    h = [sigmoid(sum(x[i] * w1[j][i] for i in range(2)) + b1[j]) for j in range(2)]
    output = sigmoid(sum(h[j] * w2[j] for j in range(2)) + b2)
    print(x, ":", round(output))

plt.plot(losses)
plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error")
plt.title("ANN Backpropagation Training")
plt.savefig("Exp-4-Output.png")
plt.show()
```

```python
import math
import matplotlib.pyplot as plt

X = [[1, 2], [2, 3], [3, 3], [6, 7], [7, 8], [8, 7]]
Y = ['A', 'A', 'A', 'B', 'B', 'B']

test = [5, 6]
k = 3

distances = []

for point, label in zip(X, Y):
    d = math.sqrt((point[0] - test[0]) ** 2 + (point[1] - test[1]) ** 2)
    distances.append((d, label, point))

distances.sort()
neighbors = distances[:k]

prediction = max(set(n[1] for n in neighbors), key=lambda c: sum(n[1] == c for n in neighbors))

print("Test Point:", test)
print("Nearest Neighbours:", [(n[2], n[1]) for n in neighbors])
print("Predicted Class:", prediction)

for label, marker in [('A', 'o'), ('B', 's')]:
    points = [p for p, y in zip(X, Y) if y == label]
    plt.scatter([p[0] for p in points], [p[1] for p in points], marker=marker, label=label)

plt.scatter(test[0], test[1], color='red', marker='*', s=200, label='Test Point')

for n in neighbors:
    p = n[2]
    plt.plot([test[0], p[0]], [test[1], p[1]], 'k--')

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Nearest Neighbours")
plt.legend()
plt.savefig("Exp-5-Output.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

actual = ['Yes', 'Yes', 'No', 'No', 'Yes', 'No']
predicted = ['Yes', 'Yes', 'No', 'Yes', 'Yes', 'No']

labels = ['Yes', 'No']
matrix = [[0, 0], [0, 0]]

# Compute confusion matrix
for a, p in zip(actual, predicted):
    matrix[labels.index(a)][labels.index(p)] += 1

# Calculate final accuracy
accuracy = sum(matrix[i][i] for i in range(2)) / len(actual) * 100

print("Confusion Matrix:", matrix)
print("Accuracy:", accuracy, "%")

# Visualize Confusion Matrix
plt.imshow(matrix, cmap='Blues')
plt.xticks([0, 1], labels)
plt.yticks([0, 1], labels)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Naive Bayes Confusion Matrix")

# Add matrix counts inside the plot
for i in range(2):
    for j in range(2):
        plt.text(j, i, matrix[i][j], ha='center', va='center', color='black')

plt.colorbar()
plt.savefig("Exp-6-Output.png")
plt.show()
```

```python
import math
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5, 6]
Y = [0, 0, 0, 1, 1, 1]

w = 1.0
b = -3.5
lr = 0.1

def sigmoid(z):
    return 1 / (1 + math.exp(-z))

# Training loop
for epoch in range(1000):
    dw = 0
    db = 0
    for x, y in zip(X, Y):
        z = w * x + b
        p = sigmoid(z)
        dw += (p - y) * x
        db += p - y

    # Update parameters outside the training samples loop
    w -= lr * dw / len(X)
    b -= lr * db / len(X)

print("Final Weight:", w)
print("Final Bias:", b)

# Generate points for the logistic regression curve
values = [i / 10 for i in range(0, 71)]
probabilities = [sigmoid(w * x_val + b) for x_val in values]

# Visualization
plt.scatter(X, Y, label="Training Data", color='blue')
plt.plot(values, probabilities, label="Logistic Curve", color='red')
plt.xlabel("Input")
plt.ylabel("Probability")
plt.title("Logistic Regression")
plt.legend()
plt.savefig("Exp-7-Output.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 6]

n = len(X)
mean_x = sum(X) / n
mean_y = sum(Y) / n

numerator = sum((X[i] - mean_x) * (Y[i] - mean_y) for i in range(n))
denominator = sum((x - mean_x) ** 2 for x in X)

m = numerator / denominator
c = mean_y - m * mean_x

predictions = [m * x + c for x in X]

print("Slope:", m)
print("Intercept:", c)
print("Predictions:", predictions)

plt.scatter(X, Y, label="Actual Data")
plt.plot(X, predictions, label="Regression Line", color='red')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")
plt.legend()
plt.savefig("Exp-8-Output.png")
plt.show()
```

```python
import random
import math
import matplotlib.pyplot as plt

random.seed(10)

data = [1, 2, 2, 3, 8, 9, 10, 11]

mean1 = 2.0
mean2 = 9.0
std1 = 1.5
std2 = 1.5
weight1 = 0.5
weight2 = 0.5

def gaussian(x, mean, std):
    return (1 / (std * math.sqrt(2 * math.pi))) * math.exp(-((x - mean) ** 2) / (2 * std ** 2))

# Expectation-Maximization Algorithm
for iteration in range(30):
    # E-Step: calculate responsibilities
    r1 = []
    r2 = []
    for x in data:
        p1 = weight1 * gaussian(x, mean1, std1)
        p2 = weight2 * gaussian(x, mean2, std2)
        total = p1 + p2
        r1.append(p1 / total)
        r2.append(p2 / total)

    # M-Step: update parameters
    sum1 = sum(r1)
    sum2 = sum(r2)

    mean1 = sum(r1[i] * data[i] for i in range(len(data))) / sum1
    mean2 = sum(r2[i] * data[i] for i in range(len(data))) / sum2

    std1 = math.sqrt(sum(r1[i] * (data[i] - mean1) ** 2 for i in range(len(data))) / sum1)
    std2 = math.sqrt(sum(r2[i] * (data[i] - mean2) ** 2 for i in range(len(data))) / sum2)

    weight1 = sum1 / len(data)
    weight2 = sum2 / len(data)

print("Mean 1:", mean1)
print("Mean 2:", mean2)
print("Weight 1:", weight1)
print("Weight 2:", weight2)

x_values = [i / 10 for i in range(0, 131)]
curve1 = [weight1 * gaussian(x, mean1, std1) for x in x_values]
curve2 = [weight2 * gaussian(x, mean2, std2) for x in x_values]

# Plot actual data histogram and learned Gaussians
plt.hist(data, bins=8, density=True, alpha=0.5, label="Data")
plt.plot(x_values, curve1, label="Gaussian 1", color='red')
plt.plot(x_values, curve2, label="Gaussian 2", color='green')
plt.xlabel("Data")
plt.ylabel("Probability Density")
plt.title("Expectation-Maximization Algorithm")
plt.legend()
plt.savefig("Exp-10-Output.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [
    [20, 10, 'Good'],
    [25, 15, 'Good'],
    [30, 20, 'Good'],
    [50, 60, 'Poor'],
    [60, 70, 'Poor'],
    [70, 80, 'Poor']
]

test = [35, 25]
k = 3

distances = []
for row in data:
    d = ((row[0] - test[0]) ** 2 + (row[1] - test[1]) ** 2) ** 0.5
    distances.append((d, row[2]))

distances.sort()
neighbors = distances[:k]

prediction = max(set(n[1] for n in neighbors),
                 key=lambda c: sum(n[1] == c for n in neighbors))

print("Test Data:", test)
print("Predicted Credit Score:", prediction)

for label, marker in [('Good', 'o'), ('Poor', 's')]:
    points = [r for r in data if r[2] == label]
    plt.scatter([r[0] for r in points],
                [r[1] for r in points],
                marker=marker, label=label)

plt.scatter(test[0], test[1], color='red', marker='*', s=200, label='New Customer')
plt.xlabel("Income")
plt.ylabel("Debt Ratio")
plt.title("Credit Score Classification using KNN")
plt.legend()
plt.savefig("Exp-11-Output.png")
plt.show()
```
